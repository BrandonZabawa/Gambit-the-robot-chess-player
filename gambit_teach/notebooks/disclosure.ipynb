## The tension

* **Constructive smurfing** only works if Gambit plays below full strength and aims at weaknesses.
* If the player *knows*, a win may feel "given" and a loss may feel "set up"; they might also play
  differently (e.g. take more risks), which moves them away from their history (OOD, see
  `player_model.ipynb`).
* If the player *does not know* and later finds out, trust may suffer ("patronizing").
* Related open questions: should Gambit ever play at full strength to avoid losing? What are the
  minimum and maximum strengths it will play at?

## Options (to evaluate)

| Option | Description | Pros | Cons |
|---|---|---|---|
| A. Always disclose up front | "I adapt to your habits and play below full strength." | honest, simple consent | may change behaviour |
| B. Disclose after the game | post-game summary: level played, habits targeted | honest, teaching moment, no in-game bias | surprise once |
| C. Disclose on request | the player can ask at any time | player control | many never ask |
| D. Never disclose | silent adaptation | no behavioural bias | trust risk; likely unacceptable for research consent |
| E. Gradual reveal | reveal more as the player improves (README: "only reveals more of its real ability as the player improves") | matches the design story | complex to explain |

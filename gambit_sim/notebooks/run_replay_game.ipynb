# Run: replay a PGN game in Gazebo

Teleports pieces ply by ply ("vanishing pieces") so you can watch a game in the sim, test game logic
against ground truth, and check frames by eye. Needs the sim running
(`ros2 launch gambit_bringup gambit.launch.py`) and `gz` on PATH.

```python
%run ./teleport.ipynb
import time

PGN = """[Event "Example"]
1. e4 e5 2. Nf3 Nc6 3. Bb5 a6 4. Ba4 Nf6 5. O-O Be7 6. Re1 b5 7. Bb3 d6 8. c3 O-O 9. h3 *"""
PAUSE_S = 1.0
DRY_RUN = True       # set False with Gazebo running
```

```python
backend = DryRunBackend() if DRY_RUN else GzCliBackend()
plies = pgn_replay(PGN)
for ply in plies:
    backend.send_all(ply.requests)
    print(f"{ply.idx:3d} {ply.san:8s} {len(ply.requests)} req  {ply.fen}")
    if not DRY_RUN:
        time.sleep(PAUSE_S)
```
